```python
# Set untuk menyimpan daftar unik jurusan mahasiswa
jurusan_a = {"Agroindustri", "Biosistem", "Agroteknologi"}
jurusan_b = {"Biosistem", "Peternakan", "Agroteknologi"}
# Menambahkan elemen baru ke dalam set
jurusan_a.add("Ilmu Tanah")
# Operasi gabungan (union) antara dua set
gabungan = jurusan_a.union(jurusan_b)
# Operasi irisan (intersection) antara dua set
irisan = jurusan_a.intersection(jurusan_b)
print(f"Gabungan: {gabungan}")
print(f"Irisan: {irisan}")
```

```python
# Fungsi dengan parameter dan return value
def hitung_total_harga(harga_per_kg, jumlah_kg):
    total = harga_per_kg * jumlah_kg
    return total
# Memanggil fungsi dan menyimpan hasilnya ke variabel
total_belanja = hitung_total_harga(15000, 8)
print(f"Total harga: Rp{total_belanja}")
```

```python
# Fungsi dengan default argument
def hitung_diskon(harga, persen_diskon=10):
  potongan = harga * (persen_diskon / 100)
  harga_akhir = harga - potongan
  return harga_akhir
# Memanggil fungsi tanpa argument kedua (menggunakan default)
print(hitung_diskon(100000))
# Memanggil fungsi dengan keyword argument
print(hitung_diskon(harga=100000, persen_diskon=25))
```

```python
# Implementasi Clean Code: Parameter dan Return Value (Tanpa keyword 'global')
def kurangi_stok(stok_saat_ini, jumlah_terjual):
  """
  Fungsi murni untuk menghitung sisa stok komoditas.
  Fungsi ini bersifat deterministik dan tidak memicu efek samping
  (side effects) pada state atau variabel di luar cakupannya.
  """
  if jumlah_terjual > stok_saat_ini:
    return stok_saat_ini # Kembalikan nilai awal jika stok tidak cukup
  sisa = stok_saat_ini - jumlah_terjual
  return sisa
# Program Utama (Pengelolaan State Berada di Luar Fungsi)
stok_gudang = 500
print(f"Stok awal sistem: {stok_gudang} unit")
# Mengubah nilai variabel dengan menangkap nilai kembalian fungsi
stok_gudang = kurangi_stok(stok_gudang, 120)
print(f"Sisa stok di gudang setelah mutasi: {stok_gudang} unit")
```

```python
def input_data_panen():
# Simulasi data hasil panen dari beberapa petak lahan
  return [120, 95, 150, 80, 110]
def hitung_total_panen(data_panen):
  return sum(data_panen)
def hitung_rata_rata_panen(data_panen):
  return hitung_total_panen(data_panen) / len(data_panen)
def tampilkan_laporan(data_panen):
  total = hitung_total_panen(data_panen)
  rata_rata = hitung_rata_rata_panen(data_panen)
  print(f"Total hasil panen: {total} kg")
  print(f"Rata-rata hasil panen: {rata_rata:.2f} kg per petak")
def main():
  data_panen = input_data_panen()
  tampilkan_laporan(data_panen)
main()
```
